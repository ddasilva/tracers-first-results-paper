```python
import lib_eventvis
import pandas as pd
from datetime import timedelta

df = pd.read_csv('output/Sept30_Storm_filtered.csv', parse_dates=['start_time', 'end_time'], index_col=0)
```

```python
def get_d(i):
    with open(f'output/Sept30_Storm/dprime_{i}.txt', 'r') as fh:
        d = float(fh.read())
        print(f'd = {d:.2f}')
        return d
```

```python
idx = 3
print(df.iloc[idx])

fig = lib_eventvis.make_event_vis(
    row=df.iloc[idx],
    ace_file='data/Sept30_Storm/ace/ts2_l2_ace_def_20250930_v1.1.0.cdf',
    scoring_file='output/Sept30_Storm/data/TracersDispersionEvent_TS2_2025-09-30T11:28:56_2025-09-30T11:30:19_SCORE.csv',
    d=get_d(idx),
    title_label='Sept 30, 2025 Storm Dst Minimum',
    dispersion_stime=df.iloc[idx].start_time + timedelta(seconds=10),
    dispersion_etime=df.iloc[idx].end_time-timedelta(seconds=17),
    eic_legend_loc='lower right',
    ascending=True,
)
fig.savefig('plots/event_sept30_dstmin.png', dpi=300, bbox_inches='tight')
```
