```python
from datetime import timedelta
import pandas as pd
import numpy as np
import lib_dasilva2026
```

```python
storm = 'Jan19_Storm'
```

```python
def get_d(i):
    with open(f'output/{storm}/dprime_{i}.txt', 'r') as fh:
        d = float(fh.read())
        print(f'd = {d:.2f}')
        return d
```

```python
def get_alpha(i):
    df_alpha = pd.read_csv('output/Jan19_Storm//alpha.csv')
    return df_alpha.iloc[i].alpha
```

```python
df = pd.read_csv(f'output/{storm}_filtered.csv', parse_dates=['start_time', 'end_time'], index_col=0)
df['i'] = np.arange(len(df))
```

```python
print(len(df))
```

# Event #0

```python
row = df.iloc[0]

data = lib_dasilva2026.load_data(
    aci_file=row.aci_file,
    ead_file=row.ead_file,
)
subset_data = data.subset(row.start_time, row.end_time)
dispersion_stime = row.start_time + timedelta(seconds=25)
dispersion_etime = row.end_time - timedelta(seconds=40)

fig, (ax1, ax2) = subset_data.plot_recon_rate(
    dispersion_stime, 
    dispersion_etime, 
    d=get_d(row.i),
    alpha=get_alpha(row.i),
    data_file=f'output/{storm}/recon_event{row.i}.csv',
    ascending=False,
)
```
