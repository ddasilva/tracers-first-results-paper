```python
import lib_eventvis
import pandas as pd
from datetime import timedelta

df = pd.read_csv('output/Jan19_Storm_filtered.csv', parse_dates=['start_time', 'end_time'], index_col=0)
```

```python
def get_d(i):
    with open(f'output/Jan19_Storm/dprime_{i}.txt', 'r') as fh:
        d = float(fh.read())
        print(f'd = {d:.2f}')
        return d
```

```python
def get_alpha(i):
    df_alpha = pd.read_csv('output/Jan19_Storm//alpha.csv')
    return df_alpha.iloc[i].alpha
```

```python
idx = 0
print(df.iloc[idx])

fig = lib_eventvis.make_event_vis(
    row=df.iloc[idx],
    ace_file='data/Jan19_Storm/ace/ts2_l2_ace_def_20260120_v1.1.0.cdf',
    scoring_file='output/Jan19_Storm/data/TracersDispersionEvent_TS2_2026-01-20T19:06:24_2026-01-20T19:07:20_SCORE.csv',
    d=get_d(idx),
    alpha=get_alpha(idx),
    offset_timedelta=timedelta(seconds=18),
    title_label='Jan 19, 2026 Storm: Event #1',
    dispersion_stime=df.iloc[idx].start_time + timedelta(seconds=25),
    dispersion_etime=df.iloc[idx].end_time-timedelta(seconds=40),
    eic_legend_loc='lower right',
    ascending=False,
)
fig.savefig('plots/event_jan19_event0.png', dpi=300, bbox_inches='tight')
```
